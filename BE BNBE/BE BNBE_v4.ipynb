```python
# %%

#------------------------------------------------ Begin_Fouction ----------------------------------------

def bourange_same_length_array(sqldict) :

    len_value=[]

    for key, value in sqldict.items():

        len_value.append(len(value))

    maxlen = max(len_value)

    for key, val in sqldict.items():

        if len(sqldict[key]) != maxlen:

            empty = []

            total_empty = maxlen - len(sqldict[key])

            for i in range(total_empty):

                empty.append('')

            sqldict[key]=sqldict[key]+empty

    return sqldict


def click_on_cookies(web_driver):

    try:

        web_driver.find_element(By.XPATH,'//*[@id="popup-buttons"]/button[1]').click()

    except Exception as err:

        print('[ERROR] : Failed to click "Allow all" button on the cookies banner:', err)


def pick_col(df, keyword):
    matches = [c for c in df.columns if keyword.lower() in c.lower()]
    return matches[0] if matches else None

def check_dowload_files(tempfolder, fileType, wait_time=10):

    for time in range(wait_time):

        if len([ele for ele in os.listdir(tempfolder) if '.crdownload' not in ele and '.tmp' not in ele]) != 0 :

            print(f"[INFO] : {fileType} file = {os.listdir(tempfolder)})")

            break

        else:

            print(f"[INFO] : Download {fileType} file ... (wait {time*2}/20 s)")

            sleep(2)

    else:

        raise Exception(f'[ERROR] : Failed to Download {fileType} file. Run Script again' )

    return  os.listdir(tempfolder)[0]



```

```python


# %%

#------------------------------------------------ Begin_writer and save df to excel  ----------------------------------------

os.chdir(scriptfolder)

df=pd.DataFrame(sqldict)



# drop duplicates and Revoked company

df = df.drop_duplicates(subset = ["Name", "ListCode"], keep = 'first')

df = df.reset_index(drop=True)

# ------ ListLabel : 1 = bank, 2 = insurance, 3 = bank & insurance, 4 = everything else ------
# Assigned per list from its ListName. Credit institutions / identification codes -> 1,
# insurance & reinsurance -> 2, investment firms, payment / e-money institutions,
# CSDs, custody and mutual guarantee companies -> 4.
ListLabelMap = {
    1: 1,   # Financial institutions in order of identification
    2: 1,   # Credit institutions authorised in Belgium
    3: 1,   # EEA credit institutions with a branch in Belgium
    4: 1,   # EEA credit institutions - freedom to provide services
    5: 1,   # Representative offices of EEA credit institutions
    6: 2,   # Licensed insurance companies
    7: 2,   # Branches of EEA insurance undertakings
    8: 2,   # EEA insurance undertakings - freedom to provide services
    9: 2,   # Reinsurance undertakings
    11: 4,  # Stockbroking firms under Belgian law
    12: 4,  # Branches of EEA stockbroking firms
    13: 4,  # Payment institutions under Belgian law
    14: 4,  # EEA payment institutions with a branch in Belgium
    15: 4,  # EEA payment institutions - freedom to provide services
    16: 4,  # Electronic money institutions authorised in Belgium
    17: 4,  # EEA e-money institutions with a branch in Belgium
    18: 4,  # EEA e-money institutions - freedom to provide services
    20: 2,  # US reinsurers operating without a branch
    21: 2,  # Insurers with full/systematic reinsurance conventions
    22: 4,  # Central securities depositories
    23: 4,  # Institutions holding dematerialised securities accounts
    24: 4,  # Account keepers for dematerialised securities
    25: 4,  # Limited-network payment service providers
    26: 4,  # Account-aggregator payment institutions
    27: 4,  # Mutual guarantee companies
}

df['ListLabel'] = df['ListCode'].astype(int).map(ListLabelMap)

# ------ ListLanguage : every source page is scraped from the /fr/ tree of nbb.be ------
df['ListLanguage'] = 'FR'

# ------ Cntry : ListCode 1 (bank identification codes) is published as a bare
# name/code XLSX with no address column. It is the Belgian national identification
# register, so every institution on it is Belgian by construction. ------
df.loc[(df['ListCode'].astype(int) == 1) & (df['Cntry'].fillna('') == ''), 'Cntry'] = 'BE'

missing = df[df['ListLabel'].isna()]['ListCode'].unique()
if len(missing):
    raise Exception('[ERROR] : - ListCode(s) with no ListLabel mapping: {}'.format(sorted(missing)))





df.to_excel(os.path.join(scriptfolder, filename), sheet_name='SQL Ready', index=False)

print('Saved {} rows to {}'.format(len(df), os.path.join(scriptfolder, filename)))

print(df.groupby('ListCode').size())

driver.quit()

sleep(3)
```
