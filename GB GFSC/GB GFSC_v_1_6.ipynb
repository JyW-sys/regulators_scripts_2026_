```python
# %%

#------------------------------------------------ Begin_Librairie ----------------------------------------

import pandas as pd

from bs4 import BeautifulSoup

from time import sleep

from datetime import datetime

from pandas import ExcelWriter

from selenium import webdriver

from selenium.webdriver.common.by import By

import datetime

import os

from selenium.webdriver.chrome.service import Service as ChromeService



# %%
```

```python
# %%

#------------------------------------------------ Begin_ fileName ----------------------------------------

print("Running GB GFSC Web Scraping Tool v.1.6")

regulatorName = 'GB GFSC'

now=datetime.datetime.now()

filename= 'GB GFSC Data {}.xlsx'.format(str(now).replace(":",".")[:-7])

scriptfolder = f"C:\\Users\\wuj1\\OneDrive - moodys.com\\Desktop\\Regulator\\{regulatorName}" ## to comment for the local environment

#scriptfolder=os.path.dirname(os.path.abspath(__file__)) ## to decomment for the production environment

os.chdir(scriptfolder)

writer = ExcelWriter(filename)

tempfolder=os.path.join(scriptfolder, 'tempfolder') 



if os.path.exists(tempfolder):

    for rem in os.listdir(tempfolder):

        os.remove(os.path.join(tempfolder, rem))

else:

    os.mkdir(tempfolder)
```

```python
# %%

#------------------------------------------------ Begin_chromedriver ----------------------------------------

#Starting Chrome driver, set to download files in tempfolder

chromeOptions = webdriver.ChromeOptions()

prefs = {"plugins.always_open_pdf_externally": True,

		 "download.prompt_for_download": False,

		 "download.default_directory" : tempfolder}

chromeOptions.add_experimental_option("prefs",prefs)

driver = webdriver.Chrome(options=chromeOptions)

driver.maximize_window()



# %%
```

```python
# %%

#------------------------------------------------ Begin_Variable ----------------------------------------

regdict={

        'GB GFSC 1': 'https://www.gfsc.gg/commission/registered-entities',

        'GB GFSC 2': 'https://www.gfsc.gg/commission/regulated-entities',
        
        'GB GFSC 3': 'https://www.gfsc.gg/commission/regulated-entities'

        }



sqldict = {'bvdid': [], 'priority': [], 'ListLabel': [], 'Typology': [], 'EntryType': [], 'Name': [], 'InternalID_1': [], 'InternalID_1_type': [], 'InternalID_2': [], 

         'InternalID_2_type': [], 'InternalID_3': [], 'InternalID_3_type': [], 'CoType': [], 'License_Type': [], 'Address_1': [], 'Address_2': [], 'City': [], 

         'Zip': [], 'Cntry': [], 'Phone': [], 'Fax': [], 'Website': [], 'Email': [], 'RegulationType': [], 'RegulationTypeCode': [], 'RegulationDate': [], 'CancellationDate': [], 

         'RegCtry': [], 'RegCode' : [], 'ListCode': [], 'ListLanguage': [], 'ListValidityDate': [], 'ListName': [], 'ListProcessDate': [], 'LEI Code': [], 'BIC SWIFT Code': [], 'Name - Mother Company': [],

         'Address_1 - Mother company': [], 'Address_2 -  Mother company': [], 'City - Mother company': [], 'Zip - Mother company': [], 'Cntry - Mother company': [], 

         'Phone - Mother company': [], 'Check': []}



processdate = now.strftime('%Y-%m-%d')


```

```python
# %%

#------------------------------------------------ Begin_Fouction ----------------------------------------

def bourange_same_length_array(sqldict) :

    maxlen = len(sqldict['ListProcessDate'])

    for key, val in sqldict.items():

        if len(sqldict[key]) != maxlen:

            empty = []

            total_empty = maxlen - len(sqldict[key])

            for i in range(total_empty):

                empty.append('')

            sqldict[key]=sqldict[key]+empty

    return sqldict



def click_on_cookies(web_driver):

    try:

        web_driver.find_element(By.XPATH,f'//*[@id="ccc-notify-accept"]/span[contains(text(),"Accept")]').click()

    except Exception as err:

        print('[ERROR] : Failed to click "I Accept" button on the cookies banner:', err)
        

# Define a function to scroll to the bottom of the page

def scroll_to_bottom(driver):

    # Get scroll height

    last_height = driver.execute_script("return document.body.scrollHeight")



    while True:

        # Scroll down to the bottom

        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")



        # Wait to load the page

        sleep(1)



        # Calculate new scroll height and compare with last scroll height

        new_height = driver.execute_script("return document.body.scrollHeight")

        if new_height == last_height:

            break

        last_height = new_height

def click_element_by_xpath(driver, xpath):

    # Find the element and click

    element = driver.find_element(By.XPATH, xpath)

    element.click()


```

```python
# %%

#------------------------------------------------ Begin_Main ----------------------------------------

for k, reg in enumerate(regdict):    

    print(f"[INFO] : Working {k+1}/{len(regdict)} _({reg})_ ")

    data_dict = {'GFSC_code':[], 'Name':[], 'Licence':[], 'Link':[]}



    driver.get(regdict[reg])

    sleep(3)

    soup = BeautifulSoup(driver.page_source,'html.parser')

    sleep(3)



    if soup.find('div', {'id':'ccc-notify'}):

        click_on_cookies(driver)

    if k ==2:
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/button') 
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/ul/li[2]/a')
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/div/div[1]/label[8]/input')
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/div/div[1]/label[10]/input')
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/div/div[3]/label[5]/input')
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/div/div[3]/label[6]/input')
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/div/div[3]/label[7]/input')
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/div/div[2]/label[3]/input')
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/div/div[2]/label[5]/input')
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/div/div[2]/label[7]/input')
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/div/div[2]/label[8]/input')
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/div/div[4]/label[3]/input')
            click_element_by_xpath(driver,'//*[@id="block-companyfeed"]/div/div[2]/div/div[6]/label[7]/input')
            
            sleep(3)
            
    scroll_to_bottom(driver)
    
    sleep(3)
    
    MultiPage = True

    pages =[]

    pagination = soup.find('div', {'class': 'company-feed_paginate'})
    

    
    try:
        
        print(len(pagination))
    
    except:
        
        driver.get(regdict[reg])
        
        sleep(3)

        soup = BeautifulSoup(driver.page_source,'html.parser')
    
        sleep(3)
        
        if soup.find('div', {'id':'ccc-notify'}):

            click_on_cookies(driver)
        
        MultiPage = True

        pages =[]

        pagination = soup.find('div', {'class': 'dataTables_paginate paging_simple_numbers'})
        
    
            
    if len(pagination)>0:

        for page in pagination.find_all('a'):

            try:

                num = int(page.text)

                pages.append(num)

            except:

                #print(page)

                continue

       

        #total_pages=max(pages)
        for span in pagination.find_all('span'):
            a = span.find_all('a')
            if a != []:
                a_txt = a[-1].text
        total_pages = a_txt
        
        

    else:

        MultiPage = False

        total_pages=1

    

    for page in range(int(total_pages)):

        if MultiPage :

            print(f"[INFO] : - Scrapping page {page+1}/{total_pages} | {reg}")



            for time in range(20):

                try:

                    NextPage = driver.find_element(By.XPATH, f'//*[@class="dataTables_paginate paging_simple_numbers"]//*/a[contains(text(),"{page+1}")]')

                    NextPage.click()

                    sleep(0.5)

                    break

                except:

                    print(f"[ERROR] : - Retrying {time+1}/20 to click bouton Number '{page+1}' ")

                    sleep(2)

            else:

                raise Exception(f'[ERROR] : - Failed to clik bouton Number  {page+1}. Run Script again' )


            
        soup = BeautifulSoup(driver.page_source,'html.parser')
        
        

        table = soup.find('table', {'class':'dataTable'})

        trs = table.find_all('tr', {'class':'odd'}) + table.find_all('tr', {'class':'even'})

        for tr in trs :

            data_dict['GFSC_code'].append( tr.find('td', {'class':'ref dtr-control'}).text )

            data_dict['Name'].append( tr.find('td', {'class':'name sorting_1'}).text )

            data_dict['Licence'].append( tr.find('td', {'class':'licences'}).text )

            data_dict['Link'].append('https://www.gfsc.gg'+ tr.find('td', {'class':'view'}).find('a')['href'] )



        if page == 2 :

            break



    for i, link in enumerate(data_dict['Link']) :

        print(f"[INFO] : -- webPage {i+1}/{len(data_dict['Link'])} | {reg} ")

        driver.get(link)

        sleep(0.5)

        soup = BeautifulSoup(driver.page_source,'html.parser')

        try:

            Address = soup.find('div', {'id':'block-gfsc-theme-content'}).find('div', {'class':'details-container'}).find_all('div', {'class':'value'})[0].text
            

        except:

            Address = ''
            
        if int(reg.split()[2])==1:
            sqldict['RegulationType'].append("Registered")
            sqldict['ListCode'].append(reg.split()[2])
        elif int(reg.split()[2])==2:
            sqldict['RegulationType'].append("Regulated")
            sqldict['ListCode'].append(reg.split()[2])
        else:
            sqldict['RegulationType'].append("Formerly Regulated")
            sqldict['ListCode'].append('2')

        

        sqldict['Name'].append(data_dict['Name'][i])

        #sqldict['Typology'].append(data_dict['Licence'][i])

        sqldict['InternalID_1'].append(data_dict['GFSC_code'][i])

        sqldict['InternalID_1_type'].append('GFSC code')

        
        
        
        #sqldict['RegulationType'].append('Supervised')

        sqldict['Address_1'].append(Address)

        sqldict['ListProcessDate'].append(processdate)

        sqldict['RegCtry'].append(reg.split()[0])
        
        sqldict['Cntry'].append(reg.split()[0])

        sqldict['RegCode'].append(reg.split()[1])

        # sqldict['ListCode'].append(reg.split()[2])
        
        if (Address!=''):
            
            sqldict['Zip'].append(Address.split(',')[-1])
            
            sqldict['City'].append(Address.split(',')[-2])
        else :
            sqldict['Zip'].append('')
            
            sqldict['City'].append('')



        if i == 10 :

            break



    sqldict = bourange_same_length_array(sqldict)


```
