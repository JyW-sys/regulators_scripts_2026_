```python
# %%

#------------------------------------------------ Begin_Librairie ----------------------------------------
import datetime
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from time import sleep
from googletrans import Translator
translator = Translator()
import pdfplumber
import os


```

```python
# %%

#------------------------------------------------ Begin_ fileName ----------------------------------------

regulatorName = 'MM CBM' ## change to current controller name

print(f"Running {regulatorName} Web Scraping Tool v.1.1")
now=datetime.datetime.now()
filename = '{} SQL Ready {}.xlsx'.format(regulatorName, str(now).replace(":",".")[:-7])

scriptfolder = f"C:\\Users\\wuj1\\OneDrive - Moody's\\Desktop\\Regulator\\{regulatorName}"

#scriptfolder=os.path.dirname(os.path.abspath(__file__)) ## to decomment for the production environment
os.chdir(scriptfolder)
tempfolder=os.path.join(scriptfolder, 'tempfolder') #if files are downloaded during the process

if os.path.exists(tempfolder):
    for rem in os.listdir(tempfolder):
        os.remove(os.path.join(tempfolder, rem))
else:
    os.mkdir(tempfolder)
```

```python
# %%

#------------------------------------------------ Begin_chromedriver ----------------------------------------

#Starting Chrome driver, set to download files in tempfolder

chromeOptions = webdriver.ChromeOptions()

prefs = {"plugins.always_open_pdf_externally": True,

		 "download.prompt_for_download": False,

		 "download.default_directory" : tempfolder}

chromeOptions.add_experimental_option("prefs",prefs)

driver = webdriver.Chrome(options=chromeOptions)

driver.maximize_window()


# %%
```

```python
# %%

#------------------------------------------------ Begin_Variable ----------------------------------------

regdict={	

# Deal with the webpage

regulatorName+' 1': 'https://www.cbm.gov.mm/content/state-owned-banks', 

regulatorName+' 2': 'https://www.cbm.gov.mm/content/3144', 

regulatorName+' 3 1': 'https://www.cbm.gov.mm/content/1228', 

regulatorName+' 3 2': 'https://www.cbm.gov.mm/content/6093', 

regulatorName+' 4': 'https://www.cbm.gov.mm/content/2653', 

regulatorName+' 5': 'https://www.cbm.gov.mm/content/126', 

regulatorName+' 6': 'https://www.cbm.gov.mm/content/46', 

regulatorName+' 7': 'https://www.cbm.gov.mm/content/6049', 

regulatorName+' 8': 'https://www.cbm.gov.mm/content/6954', 


}

Typology={
regulatorName+' 1':	'State-Owned Banks',
regulatorName+' 2':	'List of Credit Bureau',
regulatorName+' 3 1': 'List of Foreign Banks',
regulatorName+' 3 2': 'List of Foreign Banks',
regulatorName+' 4':	'List of Mobile Financial Services Providers',
regulatorName+' 5':	'List of Non-Bank Financial Institutions',
regulatorName+' 6':	'List of Private Banks',
regulatorName+' 7':	'List of Remittance Business Licensees',
regulatorName+' 8':	'List of Representative Offices of Foreign Banks and Finance Companies',

}

sqldict={'bvdid': [], 'priority': [], 'ListLabel': [], 'Typology': [], 'EntryType': [], 'Name': [], 'InternalID_1': [], 'InternalID_1_type': [], 'InternalID_2': [], 

	  'InternalID_2_type': [], 'InternalID_3': [], 'InternalID_3_type': [], 'CoType': [], 'License_Type': [], 'Address_1': [], 'Address_2': [], 'City': [], 

		  'Zip': [], 'Cntry': [], 'Phone': [], 'Fax': [], 'Website': [], 'Email': [], 'RegulationType': [], 'RegulationTypeCode': [], 'RegulationDate': [], 'CancellationDate': [], 

		  'RegCtry': [], 'RegCode' : [], 'ListCode': [], 'ListLanguage': [], 'ListValidityDate': [], 'ListName': [], 'ListProcessDate': [], 'LEI Code': [], 'BIC SWIFT Code': [], 'Name - Mother Company': [],

		  'Address_1 - Mother company': [], 'Address_2 -  Mother company': [], 'City - Mother company': [], 'Zip - Mother company': [], 'Cntry - Mother company': [], 

		  'Phone - Mother company': [], 'Check': []}



processdate = now.strftime('%Y-%m-%d')


```

```python
#------------------------------------------------ Begin_Fouction ----------------------------------------

def bourange_same_length_array(sqldict) :

    len_value=[]

    for key, value in sqldict.items():

        len_value.append(len(value))

    maxlen = max(len_value)

    for key, val in sqldict.items():

        if len(sqldict[key]) != maxlen:

            empty = []

            total_empty = maxlen - len(sqldict[key])

            for i in range(total_empty):

                empty.append('')

            sqldict[key]=sqldict[key]+empty

    return sqldict
```

```python
#------------------------------------------------ Begin_Fouction ----------------------------------------

def bourange_same_length_array(sqldict) :

    len_value=[]

    for key, value in sqldict.items():

        len_value.append(len(value))

    maxlen = max(len_value)

    for key, val in sqldict.items():

        if len(sqldict[key]) != maxlen:

            empty = []

            total_empty = maxlen - len(sqldict[key])

            for i in range(total_empty):

                empty.append('')

            sqldict[key]=sqldict[key]+empty

    return sqldict
```

```python
# %%

#------------------------------------------------ Begin_Main ----------------------------------------

for k, reg in enumerate(regdict):    

	print(f"[INFO] : Working {k+1}/{len(regdict)} _({reg})_ ")
	driver.get(regdict[reg])

	
	if reg == regulatorName + ' 1':
		table = driver.find_element(By.TAG_NAME,'table')
		for row in table.find_elements(By.TAG_NAME,'tr')[1:]:
			for index,td in enumerate(row.find_elements(By.TAG_NAME,'td')):
				if index ==1:
					name_ = td.text
					sqldict['Name'].append(name_)
					sqldict['ListProcessDate'].append(processdate)
					sqldict['RegCtry'].append(reg.split(' ')[0])
					sqldict['RegCode'].append(reg.split(' ')[1])
					sqldict['ListCode'].append(reg.split(' ')[2])
					sqldict['RegulationType'].append('Regulated')
					sqldict['ListName'].append(Typology[reg]) 
				elif index == 2:
					address_ = td.text
					sqldict['Address_1'].append(address_)
				
	elif reg == regulatorName + ' 2' or reg == regulatorName + ' 4' or  reg == regulatorName + ' 5' or  reg == regulatorName + ' 6':
		table = driver.find_element(By.TAG_NAME,'table')
		for row in table.find_elements(By.TAG_NAME,'tr')[1:]:
			for index,td in enumerate(row.find_elements(By.TAG_NAME,'td')):
				if index ==1:
					name_ = td.text
					sqldict['Name'].append(name_)
					sqldict['ListProcessDate'].append(processdate)
					sqldict['RegCtry'].append(reg.split(' ')[0])
					sqldict['RegCode'].append(reg.split(' ')[1])
					sqldict['ListCode'].append(reg.split(' ')[2])
					sqldict['RegulationType'].append('Regulated')
					sqldict['ListName'].append(Typology[reg]) 
				elif index == 2:
					date_licence_ = td.text
					sqldict['RegulationDate'].append(date_licence_)
				elif index == 3:
					address_ = td.text
					sqldict['Address_1'].append(address_)
				

	elif reg == regulatorName + ' 3 1' or  reg == regulatorName + ' 3 2':
		table = driver.find_element(By.TAG_NAME,'table')
		for row in table.find_elements(By.TAG_NAME,'tr')[1:]:
			for index,td in enumerate(row.find_elements(By.TAG_NAME,'td')):
				if index ==1:
					name_ = td.text
					sqldict['Name'].append(name_)
					sqldict['ListProcessDate'].append(processdate)
					sqldict['RegCtry'].append(reg.split(' ')[0])
					sqldict['RegCode'].append(reg.split(' ')[1])
					sqldict['ListCode'].append(reg.split(' ')[2])
					sqldict['RegulationType'].append('Regulated')
					sqldict['ListName'].append(Typology[reg]) 
				elif index == 2:
					date_licence_ = td.text.split('/')[0]
					sqldict['RegulationDate'].append(date_licence_)
				elif index == 3:
					address_ = td.text
					sqldict['Address_1'].append(address_)
				
 
	elif reg == regulatorName + ' 7':
		driver.find_element(By.LINK_TEXT,'List of Remittance Business Licensees').click()
		sleep(3)
		tables = []
		print(os.listdir(tempfolder))
		pdf_file = os.listdir(tempfolder)[0]
		filePath = os.path.join(tempfolder, pdf_file)     
		with pdfplumber.open(filePath) as pdf:
			for page in pdf.pages:
				table = page.extract_table()
				tables.append(table)
		for table in tables:
			if table:
				for tab in table[1:]:
					name_ = tab[1]
					date_licence_ = translator.translate(tab[2], dest='en').text
					address_  = translator.translate(tab[3], dest='en').text
					# print(name_,date_registered.text,address_.text)
					sqldict['Name'].append(name_.replace('\n',' '))
					sqldict['ListProcessDate'].append(processdate)
					sqldict['RegCtry'].append(reg.split(' ')[0])
					sqldict['RegCode'].append(reg.split(' ')[1])
					sqldict['ListCode'].append(reg.split(' ')[2])
					sqldict['Address_1'].append(address_.replace('\n',' '))
					sqldict['Address_1 - Mother company'].append(tab[3].replace('\n',' '))
					sqldict['RegulationDate'].append(date_licence_)
					sqldict['RegulationType'].append('Regulated')
					sqldict['ListName'].append(Typology[reg]) 
	elif reg == regulatorName + ' 8':
		tables = driver.find_elements(By.TAG_NAME,'table')
		for table in tables:
			for row in table.find_elements(By.TAG_NAME,'tr')[1:]:
				for index,td in enumerate(row.find_elements(By.TAG_NAME,'td')):
					if index ==1:
						name_ = td.text
						sqldict['Name'].append(name_.replace('\n',' '))
						sqldict['ListProcessDate'].append(processdate)
						sqldict['RegCtry'].append(reg.split(' ')[0])
						sqldict['RegCode'].append(reg.split(' ')[1])
						sqldict['ListCode'].append(reg.split(' ')[2])
						sqldict['RegulationType'].append('Regulated')
						sqldict['ListName'].append(Typology[reg]) 
					elif index == 2:
						date_licence_ = td.text
						sqldict['RegulationDate'].append(date_licence_)
					

	sqldict = bourange_same_length_array(sqldict)
```

```python
# %%

#------------------------------------------------ Begin_writer and save df to excel  ----------------------------------------

os.chdir(scriptfolder)

df=pd.DataFrame(sqldict)

# df= df.drop_duplicates()
df["Zip"] = df["Address_1"].str.extract(r"(\d{5})", expand=False).fillna("")
df["Name"] = df["Name"].str.replace(r"\*+", "", regex=True).str.strip()

df["RegulationDate"] = (
    df["RegulationDate"]
      .str.strip()
      .str.removeprefix("(")
      .str.removesuffix(")")
      .str.strip()
)

df.to_excel(filename, index=False)

driver.quit()

sleep(3)

for rem in os.listdir(tempfolder):
    os.remove(os.path.join(tempfolder, rem))
```
