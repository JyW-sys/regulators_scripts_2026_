```python
# %%



#------------------------------------------------ Begin_Librairie ----------------------------------------



from bs4 import BeautifulSoup

from selenium import webdriver

import datetime



import pandas as pd

import re

from pandas import ExcelWriter

import requests



from time import sleep



import os

import numpy as np



# %%



#------------------------------------------------ Begin_ fileName ----------------------------------------



regulatorName = 'MT MFSA' ## change to current controller name







print(f"Running {regulatorName} Web Scraping Tool v.1.1")



now=datetime.datetime.now()



filename = '{} SQL Read {}.xlsx'.format(regulatorName, str(now).replace(":",".")[:-7])





scriptfolder = f"C:\\Users\\wuj1\\OneDrive - Moody's\\Desktop\\Regulator\\{regulatorName}"

#scriptfolder=os.path.dirname(os.path.abspath(__file__)) ## to decomment for the production environment



os.chdir(scriptfolder)



writer = ExcelWriter(filename)



tempfolder=os.path.join(scriptfolder, 'tempfolder') #if files are downloaded during the process







if os.path.exists(tempfolder):



    for rem in os.listdir(tempfolder):



        os.remove(os.path.join(tempfolder, rem))



else:



    os.mkdir(tempfolder)



# %%



#------------------------------------------------ Begin_Variable ----------------------------------------



regdict={



        'MT MFSA 1':'License Holders' , 



        }







sqldict = {'bvdid': [], 'priority': [], 'ListLabel': [], 'Typology': [], 'EntryType': [], 'Name': [], 'InternalID_1': [], 'InternalID_1_type': [], 'InternalID_2': [], 



         'InternalID_2_type': [], 'InternalID_3': [], 'InternalID_3_type': [], 'CoType': [], 'License_Type': [], 'Address_1': [], 'Address_2': [], 'City': [], 



         'Zip': [], 'Cntry': [], 'Phone': [], 'Fax': [], 'Website': [], 'Email': [], 'RegulationType': [], 'RegulationTypeCode': [], 'RegulationDate': [], 'CancellationDate': [], 



         'RegCtry': [], 'RegCode' : [], 'ListCode': [], 'ListLanguage': [], 'ListValidityDate': [], 'ListName': [], 'ListProcessDate': [], 'LEI Code': [], 'BIC SWIFT Code': [], 'Name - Mother Company': [],



         'Address_1 - Mother company': [], 'Address_2 -  Mother company': [], 'City - Mother company': [], 'Zip - Mother company': [], 'Cntry - Mother company': [], 



         'Phone - Mother company': [], 'Check': []}





processdate = now.strftime('%Y-%m-%d')



# %%



#------------------------------------------------ Begin_Fouction ----------------------------------------



def bourange_same_length_array(sqldict) :



    len_value=[]



    for key, value in sqldict.items():



        len_value.append(len(value))



    maxlen = max(len_value)



    for key, val in sqldict.items():



        if len(sqldict[key]) != maxlen:



            empty = []



            total_empty = maxlen - len(sqldict[key])



            for i in range(total_empty):



                empty.append('')



            sqldict[key]=sqldict[key]+empty



    return sqldict




def extract_last_zip_code(address):
    # Find all sequences of 4 or more digits
    # Return the last match if any, otherwise return empty string
    zip_matches = re.findall(r'\d{4,}', address)

    return zip_matches[-1] if zip_matches else ''





# %%

#------------------------------------------------ Begin_chromedriver ----------------------------------------
#Starting Chrome driver, set to download files in tempfolder



chromeOptions = webdriver.ChromeOptions()

prefs = {"plugins.always_open_pdf_externally": True,

		 "download.prompt_for_download": False,

		 "download.default_directory" : tempfolder}

chromeOptions.add_experimental_option("prefs",prefs)

chromeOptions.add_experimental_option("excludeSwitches", ["enable-automation"])

user_agent = "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"

chromeOptions.add_argument(f"user-agent={user_agent}")



driver = webdriver.Chrome(options=chromeOptions)

driver.maximize_window()

#------------------------------------------------ Begin_Main ----------------------------------------



session = requests.Session()



url = "https://fsr.mfsa.mt/LicenceTypes/getParentLicenceTypes"



headers = {

    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8,application/signed-exchange;v=b3;q=0.7",

    "Accept-Encoding": "gzip, deflate, br, zstd",

    "Accept-Language": "en-US,en;q=0.9,zh;q=0.8,zh-CN;q=0.7",

    "Cache-Control": "no-cache",

    "Content-Type": "application/x-www-form-urlencoded",

    "Origin": "https://fsr.mfsa.mt",

    "Referer": "https://fsr.mfsa.mt/",

    "Sec-Fetch-Dest": "iframe",

    "Sec-Fetch-Mode": "navigate",

    "Sec-Fetch-Site": "same-origin",

    "Sec-Fetch-User": "?1",

    "Upgrade-Insecure-Requests": "1",

    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/139.0.0.0 Safari/537.36 Edg/139.0.0.0",

   

}



token_url = "https://fsr.mfsa.mt/Licences"

response = session.get(token_url)



# Step 3: Check cookies received

# for cookie in session.cookies:

#     print(f"{cookie.name} = {cookie.value}")



# Step 4: Parse HTML to extract __RequestVerificationToken (if present)

soup = BeautifulSoup(response.text, "html.parser")

token_input = soup.find("input", {"name": "__RequestVerificationToken"})

token = token_input["value"] if token_input else None

parentLicenceTyoeName = ''





response = requests.get(url, headers=headers)







print("Status Code:", response.status_code)



# Check response status

if response.status_code == 200:

    json_data = response.json()

    # Extract keys and values



    #for item in json_data:

    # use the first one as test

    # original logic



    # for item in json_data:

    #     for key, value in item.items():



    print('Total Sector:', len(json_data))

    for item in json_data:
        for key, value in item.items():

            if key == 'parentLicenceTypeId':

                print('Parent Licence Type ID: ',value)

                parentLicenceTyoeName = next(item['parentLicenceTypeName'] for item in json_data if item['parentLicenceTypeId'] == value)

                subtype_url = f"https://fsr.mfsa.mt/LicenceTypes/getLicenceTypesByParentId?parentLicenceTypeId={value}"

                subtype_response = requests.get(subtype_url, headers=headers)

                if subtype_response.status_code == 200:

                    subtype_json = subtype_response.json()



                    for subtype in subtype_json:

                        print('Sub Type Licence Type ID: ',subtype['licenceTypeId'])

                        general_info_url = f"https://fsr.mfsa.mt/Licences/getLicenceHoldersByLicenceTypeId?licenceTypeId={subtype['licenceTypeId']}"

                        general_info_resp = requests.get(general_info_url, headers=headers)

                        if general_info_resp.status_code == 200:

                            general_info = general_info_resp.json()

                            ### Some companies without view page, name/ code/ general info have to be from general_info in otder to make sure the correct number

                            # Iterate through the JSON data

                            for item in general_info:

                                data_entry = {}  # Dictionary to store key-value pairs for each item

                                for key, value in item.items():

                                    data_entry[key] = value

                                sqldict['Name'].append(data_entry['licenceHolderName']) 

                                sqldict['ListProcessDate'].append(processdate)

                                sqldict['InternalID_1'].append(data_entry['companyId'])

                                sqldict['InternalID_1_type'].append('companyId')

                                sqldict['RegulationType'].append(data_entry['licences'][0]['licenceStatus'])

                                sqldict['RegCtry'].append('MT')

                                sqldict['RegCode'].append('MFSA')

                                sqldict['ListCode'].append('1')

                                sqldict['License_Type'].append(parentLicenceTyoeName)

                                sqldict['ListLabel'].append(data_entry['licences'][0]['licenceTypeName'])

                                sqldict['ListName'].append('License Holders')



                            

                                if 'licences' in item:

                                    for licence in item['licences']:

                                        licenceId = licence['licenceId']



                                        payload = {

                                            "__RequestVerificationToken": token,

                                            "licenceId": licenceId

                                        }



                                        post_url = "https://fsr.mfsa.mt/Licences/ViewLicence"

                                        response_post = session.post(post_url, data=payload, headers=headers)

                                        

                                        # print("POST Status:", response_post.status_code)

                                        # print('Personal Token:', token)



                                        if not str(response_post.status_code).startswith('2'):

                                            sqldict = bourange_same_length_array(sqldict)

                                        else:



                                            soup = BeautifulSoup(response_post.text, "html.parser")





                                            # Loop through all label elements

                                            for label in soup.find_all("label"):

                                                key = label.get_text(strip=True)



                                                # Step 1: Get the parent of the label

                                                parent = label.find_parent()



                                                # Step 2: Find the next sibling <div> of that parent

                                                value_div = parent.find_next_sibling("div") if parent else None



                                                # Step 3: Extract text from the value div

                                                value = value_div.get_text(strip=True) if value_div else "Not found"

                                                

                                                if 'LEI' in key:

                                                    sqldict['LEI Code'].append(value)

                                                elif 'Business Address' in key:

                                                    sqldict['Address_1'].append(value)
                                                    sqldict['Zip'].append(extract_last_zip_code(value))

                                                elif 'Authorisation Issue Date' in key:

                                                    sqldict['RegulationDate'].append(value)

                                                elif 'Phone' in key:

                                                    sqldict['Phone'].append(value)

                                                # elif 'Email' in key:

                                                #     sqldict['Email'].append(value)

                                                elif 'Website' in key:

                                                    sqldict['Website'].append(value)



                                            sqldict = bourange_same_length_array(sqldict)



                                



                else:

                    print(f"Failed to fetch subtypes for parentLicenceTypeId {value}. Status code: {subtype_response.status_code}")

else:

    print(f"Failed to fetch data. Status code: {response.status_code}")





# %%



#------------------------------------------------ Begin_writer and save df to excel  ----------------------------------------



os.chdir(scriptfolder)



df=pd.DataFrame(sqldict)

df = df[df['Name'] != '']

df.to_excel(writer, 'SQL Ready', index=False)



writer.save()



writer.close()


#Moving the file to the output folder (this way it will be displayed in the Control Room)


driver.quit()
sleep(3)


    
    
    
    
```
