```python
# %%

#------------------------------------------------ Begin_Fouction ----------------------------------------

def bourange_same_length_array(sqldict) :

    maxlen = len(sqldict['ListProcessDate'])

    for key, val in sqldict.items():

        if len(sqldict[key]) != maxlen:

            empty = []

            total_empty = maxlen - len(sqldict[key])

            for i in range(total_empty):

                empty.append('')

            sqldict[key]=sqldict[key]+empty

    return sqldict



def findZip(mydata):
    if mydata is None:
        return ''
    if pd.isna(mydata):
        return ''
    mydata = str(mydata)
    regex = re.compile(r'\b\d{5}\b')
    zip_codes = regex.findall(mydata)
    return zip_codes[-1] if zip_codes else ''

    
```

```python
session = requests.Session()

base_headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) ...",
    "Accept-Language": "en-US,en;q=0.9,zh;q=0.8,zh-CN;q=0.7",
}
session.get("https://www.bot.or.th/en/fi-list.html", headers=base_headers)




MODEL_URL = "https://www.bot.or.th/content/bot/en/fi-list/jcr:content/root/container/involvepartyopenlist.model.json"

session.get("https://www.bot.or.th/en/fi-list.html", headers=base_headers)

api_headers = {
    **base_headers,
    "Accept": "application/json",
    "Referer": "https://www.bot.or.th/en/fi-list.html?listingType=InvolvePartyOpenListingResultsBank",
}


response = session.get(MODEL_URL, headers=api_headers)
response.raise_for_status()
model_data = response.json()

inst_type_map = model_data['listingFiltersBank'][0]['filtersMap']

```

```python
BASE = "https://www.bot.or.th/content/bot/en/fi-list/jcr:content/root/container/involvepartyopenlist"

PATTERNS = {
    "InvolvePartyOpenListingResultsBank":
        "{base}.{limit}.p{offset}.{sort}.inst{inst}.paysys{paysys}.status{status}.sk0.{order}.json",
    "InvolvePartyOpenListingResultsNonBank":
        "{base}.{limit}.p{offset}.{sort}.license{license}.status{status}.sk0.{order}.json",
}

api_headers = {
    **base_headers,
    "Accept": "application/json",
    "Referer": "https://www.bot.or.th/en/fi-list.html?listingType=InvolvePartyOpenListingResultsBank",
}

def build_url(listing_type, limit, offset=0, sort="BANK_CODE", order="ascending",
              inst="0", paysys="0", status="0", license="0"):
    pattern = PATTERNS[listing_type]
    return pattern.format(
        base=f"{BASE}.{listing_type}",
        limit=limit,
        offset=offset,
        sort=sort,
        order=order,
        inst=inst,
        paysys=paysys,
        status=status,
        license=license,
    )

for key, listing_type in ListType.items():
    first_url = build_url(listing_type, limit=20)  # use 20 (UI max) for the first call
    resp = session.get(first_url, headers=api_headers, timeout=20)
    resp.raise_for_status()
    payload = resp.json()

    total_results = payload["totalResults"]
    full_url = build_url(listing_type, limit=total_results)
    resp = session.get(full_url, headers=api_headers, timeout=20)
    resp.raise_for_status()
    payload = resp.json()

    records = payload["results"]
    df = pd.json_normalize(records)
    print(key, df.shape)
    df_true = df[df['activeStatus'] == True]
    print('ActiveStatus: ', df_true.shape)

    if listing_type == 'InvolvePartyOpenListingResultsNonBank':
        for name,address, tel_, fax_, web_,email_ in zip (df_true['OrgNmEng'],df_true['addressThai'],df_true['tel'],df_true['fax'],df_true['webUrl'],df_true['email']):
            #print(name,address, tel_, fax_, web_,email_ )
            sqldict['Name'].append(name)
            sqldict['ListProcessDate'].append(processdate)
            sqldict['Address_1'].append(address)
            sqldict['Zip'].append(findZip(address))
            tel_clean = "" if pd.isna(tel_) else tel_
            sqldict['Phone'].append(tel_clean)
            fax_clean  = "" if pd.isna(fax_) else fax_
            fax_clean = '' if len(fax_clean.split(':')[-1])<3 else fax_clean.split(':')[-1]
            sqldict['Fax'].append(fax_clean)
            web_clean  = "" if pd.isna(web_) else web_
            sqldict['Website'].append(web_clean)
            email_clean  = "" if pd.isna(web_) else web_
            sqldict['Email'].append(email_clean)
            sqldict['RegulationType'].append('Regulated')
            sqldict['RegCtry'].append('TH')
            sqldict['RegCode'].append('BTHAI')
            sqldict['ListCode'].append('1')
            sqldict['ListName'].append('Institution Listing')
    else:
        for name, address, bankcode, type_id,web_,fax_,tel_ in zip(df_true['institutionName'],df_true['institutionAddress'],df_true['bankCode'],df_true['institutionTpId'],df_true['institutionUrl'],df_true['fax'],df_true['telephone']):
            topology = inst_type_map.get(type_id)
            sqldict['Typology'].append(topology)
            sqldict['Name'].append(name)
            sqldict['ListProcessDate'].append(processdate)
            sqldict['Address_1'].append(address)
            sqldict['Zip'].append(findZip(address))
            sqldict['InternalID_1'].append(bankcode)
            sqldict['InternalID_1_type'].append('BankCode')
            tel_clean = "" if pd.isna(tel_) else tel_
            sqldict['Phone'].append(tel_clean)
            fax_clean  = "" if pd.isna(fax_) else fax_
            fax_clean = '' if len(fax_clean.split(':')[-1])<3 else fax_clean.split(':')[-1]
            sqldict['Fax'].append(fax_clean)
            web_clean  = "" if pd.isna(web_) else web_
            sqldict['Website'].append(web_clean)
            sqldict['RegulationType'].append('Regulated')
            sqldict['RegCtry'].append('TH')
            sqldict['RegCode'].append('BTHAI')
            sqldict['ListCode'].append('1')
            sqldict['ListName'].append('Institution Listing')
    sqldict = bourange_same_length_array(sqldict)
```
